```python
date_window = Window.partitionBy("machine_id").orderBy("production_date", "production_id")
df_windowed = df.withColumn(
    "cumulative_production",
    F.sum("units_produced").over(date_window.rowsBetween(Window.unboundedPreceding, Window.currentRow))
).withColumn(
    "previous_output", F.lag("units_produced").over(date_window)
).withColumn(
    "next_output", F.lead("units_produced").over(date_window)
).withColumn(
    "change_from_previous",
    F.col("units_produced") - F.lag("units_produced").over(date_window)
)
df_windowed.select(
    "machine_id", "production_date", "units_produced", "cumulative_production",
    "previous_output", "next_output", "change_from_previous"
).show(20)
```

```python
df_dates.withColumn(
    "date_minus_30_days", F.date_sub("production_date", 30)
).select("production_date", "date_minus_30_days").show(10)
```

```python
df_dates.withColumn(
    "days_from_current_date", F.datediff(F.current_date(), F.col("production_date"))
).select("production_date", "days_from_current_date").show(10)
```

```python
monthly_output = df_dates.groupBy("year_month").agg(
    F.sum("units_produced").alias("total_output")
)
monthly_output.orderBy(F.col("total_output").desc()).show(1, truncate=False)
```
